```python

from collections import Counter, defaultdict
from statistics import mean

LAST_NAME, SEED_NUM, FAVORITE_ARTIST = "Medina", 5, "Kendrick Lamar"
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

data = [(s, c, (ord(FAVORITE_ARTIST[(i+j) % len(FAVORITE_ARTIST)])
        + SEED_NUM*(i+1)+j) % 100 + 1)
        for i, s in enumerate(sources)
        for j, c in enumerate(categories)]
data += [data[0], (sources[0], "TEMPERATURE", 35),
         (sources[1], "VOLTAGE", 48)]

organized = defaultdict(lambda: defaultdict(list))
for s, c, v in data:
    organized[s][c].append(v)

counts = Counter(data)
print("Generated Data:", data)
print("Organized Data:", dict(organized))
print("Repeated:", [r for r, n in counts.items() if n > 1])
print("Distinct Records:", len(counts), "| Total:", len(data))

for c in categories:
    values = [v for _, cat, v in data if cat == c]
    print(f"{c}: Count={len(values)}, Mean={mean(values):.2f}, "
          f"Min={min(values)}, Max={max(values)}")

print("Ordered Results:", sorted(data))
print("Final Summary: Sources=5, Categories=5, Records=", len(data))
```

```python

from collections import Counter, defaultdict
from statistics import mean

LAST_NAME, SEED_NUM, FAVORITE_ARTIST = "Medina", 5, "Kendrick Lamar"
students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

data = [(s, sub, (ord(FAVORITE_ARTIST[(i+j) % len(FAVORITE_ARTIST)])
        + SEED_NUM*(i+1)+j) % 51 + 50)
        for i, s in enumerate(students)
        for j, sub in enumerate(subjects)]
data += [data[0], (students[0], "ENGLISH", None)]

def category(score):
    if score is None: return "INCOMPLETE"
    if score >= 90: return "Excellent"
    if score >= 80: return "Very Good"
    if score >= 75: return "Passed"
    return "Needs Improvement"

counts = Counter(data)
print("Generated Records:", data)
print("Repeated:", [r for r, n in counts.items() if n > 1])
print("Incomplete:", [r for r in data if r[2] is None])
print("Categories:", Counter(category(r[2]) for r in data))

for label, index, names in [("Student", 0, students), ("Subject", 1, subjects)]:
    for name in names:
        scores = [r[2] for r in data if r[index] == name and r[2] is not None]
        print(f"{label} {name}: Count={len(scores)}, "
              f"Average={mean(scores):.2f}, Min={min(scores)}, Max={max(scores)}")

print("Ordered Results:", sorted(data, key=lambda r: (r[0], r[1], r[2] or 0)))
print(f"Final Report: Students=5, Subjects=5, Records={len(data)}")
```

```python

from collections import Counter, defaultdict

LAST_NAME, SEED_NUM, FAVORITE_ARTIST = "Medina", 5, "Kendrick Lamar"
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
prices = {"NOTEBOOK": 35, "PEN": 15, "USB": 250,
          "HEADPHONES": 450, "CALCULATOR": 300}
products = list(prices)

data = [(c, p, (ord(FAVORITE_ARTIST[(i+j) % len(FAVORITE_ARTIST)])
        + SEED_NUM+i+j) % 5 + 1, prices[p])
        for i, c in enumerate(customers)
        for j, p in enumerate(products)]
data.append(data[0])

records = [(c, p, q, price, q*price) for c, p, q, price in data]
pairs = Counter((c, p) for c, p, *_ in records)
customer = defaultdict(lambda: [0, 0])
product = defaultdict(lambda: [0, 0])

for c, p, q, price, total in records:
    customer[c][0] += q
    customer[c][1] += total
    product[p][0] += q
    product[p][1] += total

print("Transactions:", records)
print("Repeated Purchases:", [p for p, n in pairs.items() if n > 1])
print("Distinct Customers/Products:", len({r[0] for r in records}),
      "/", len({r[1] for r in records}))
print("Customer Summary:", dict(customer))
print("Product Summary:", dict(product))
print("Ordered Results:", sorted(records, key=lambda r: (r[0], r[1])))
print(f"Final Report: Records={len(records)}, Units={sum(r[2] for r in records)}, "
      f"Revenue=₱{sum(r[4] for r in records):,}")
```
