```python
# EXERCISE 1: Engineering Test Data Organizer


LAST_NAME = "Medina"
SEED_NUM = 5
FAVORITE_ARTIST = "Kendrick Lamar"


# Student Inputs
test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]


# 1. Generated Test Data (Raw List with duplicates & multi-measurements)
raw_test_data = [
    {"source": test_sources[0], "category": categories[0], "value": SEED_NUM * ord(FAVORITE_ARTIST[0]) % 100},
    {"source": test_sources[1], "category": categories[1], "value": SEED_NUM * ord(FAVORITE_ARTIST[1]) % 100},
    {"source": test_sources[2], "category": categories[2], "value": SEED_NUM * ord(FAVORITE_ARTIST[2]) % 100},
    {"source": test_sources[3], "category": categories[3], "value": SEED_NUM * ord(FAVORITE_ARTIST[3]) % 100},
    {"source": test_sources[4], "category": categories[4], "value": SEED_NUM * ord(FAVORITE_ARTIST[4]) % 100},
    # Duplicate record
    {"source": test_sources[0], "category": categories[0], "value": SEED_NUM * ord(FAVORITE_ARTIST[0]) % 100},
    # Extra measurements for specific categories
    {"source": test_sources[0], "category": categories[0], "value": 85.5},
    {"source": test_sources[1], "category": categories[1], "value": 12.4},
]


print("=== EXERCISE 1: ENGINEERING TEST DATA ORGANIZER ===")
print("1. Generated Test Data:")
for item in raw_test_data:
    print(item)


# 2. De-duplication (Distinct Records) using set of tuples
distinct_records = list({(d["source"], d["category"], d["value"]) for d in raw_test_data})
print(f"\n2. Total Raw Records: {len(raw_test_data)} | Distinct Records: {len(distinct_records)}")


# 3. Categorical & Numerical Summaries
category_values = {}
for src, cat, val in distinct_records:
    category_values.setdefault(cat, []).append(val)


print("\n3. Category Summaries (Average, Min, Max):")
for cat, vals in category_values.items():
    avg_val = sum(vals) / len(vals)
    print(f"   - {cat}: Count={len(vals)}, Avg={avg_val:.2f}, Min={min(vals)}, Max={max(vals)}")




# EXERCISE 2: Student Performance Data Analyzer




students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]


# Generate performance records
performance_records = [
    {"student": students[0], "subject": subjects[0], "score": (ord(FAVORITE_ARTIST[0]) + SEED_NUM * 10) % 101},
    {"student": students[1], "subject": subjects[1], "score": (ord(FAVORITE_ARTIST[1]) + SEED_NUM * 10) % 101},
    {"student": students[2], "subject": subjects[2], "score": (ord(FAVORITE_ARTIST[2]) + SEED_NUM * 10) % 101},
    {"student": students[3], "subject": subjects[3], "score": (ord(FAVORITE_ARTIST[3]) + SEED_NUM * 10) % 101},
    {"student": students[4], "subject": subjects[4], "score": None},  # Incomplete record
    # Duplicate student-subject pair
    {"student": students[0], "subject": subjects[0], "score": (ord(FAVORITE_ARTIST[0]) + SEED_NUM * 10) % 101},
]


print("\n=== EXERCISE 2: STUDENT PERFORMANCE DATA ANALYZER ===")
print("1. Generated Performance Records:")
for r in performance_records:
    print(r)


# Identify incomplete and duplicate records
valid_records = [r for r in performance_records if r["score"] is not None]
incomplete_records = [r for r in performance_records if r["score"] is None]


print(f"\n2. Valid Records: {len(valid_records)} | Incomplete Records: {len(incomplete_records)}")


# Subject performance categories
print("\n3. Student Performance Evaluation:")
for r in valid_records:
    score = r["score"]
    category = "Excellent" if score >= 85 else "Good" if score >= 70 else "Needs Improvement"
    print(f"   - {r['student']} [{r['subject']}]: Score = {score} ({category})")




# EXERCISE 3: Transaction History Analyzer


customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
fixed_prices = {"NOTEBOOK": 50.0, "PEN": 20.0, "USB": 350.0, "HEADPHONES": 500.0, "CALCULATOR": 450.0}


transactions = [
    {"customer": customers[0], "product": products[0], "quantity": (SEED_NUM + len(FAVORITE_ARTIST)) % 5 + 1},
    {"customer": customers[1], "product": products[1], "quantity": (SEED_NUM + 2) % 5 + 1},
    {"customer": customers[2], "product": products[2], "quantity": (SEED_NUM + 3) % 5 + 1},
    {"customer": customers[3], "product": products[3], "quantity": (SEED_NUM + 4) % 5 + 1},
    {"customer": customers[4], "product": products[4], "quantity": (SEED_NUM + 1) % 5 + 1},
    # Duplicate customer transaction
    {"customer": customers[0], "product": products[0], "quantity": (SEED_NUM + len(FAVORITE_ARTIST)) % 5 + 1},
]


print("\n=== EXERCISE 3: TRANSACTION HISTORY ANALYZER ===")
total_overall = 0.0


print("1. Processed Transactions:")
for t in transactions:
    unit_price = fixed_prices[t["product"]]
    total_val = t["quantity"] * unit_price
    t["total_value"] = total_val
    total_overall += total_val
    print(f"   - {t['customer']} bought {t['quantity']} x {t['product']} @ ₱{unit_price} = ₱{total_val:.2f}")


print(f"\n2. Total Overall Dataset Transaction Value: ₱{total_overall:.2f}")
```
